What will happen if we take more that two classes at a time?

If I use more than two classes at a time, the classification problem becomes more difficult because the model must distinguish among several groups instead of only class 0 and class 1.
The same features can still be used, but some classes may overlap more than others. This can reduce accuracy if the features do not clearly separate all classes.
I would use a multi-class classifier and evaluate it using overall accuracy and a confusion matrix. The confusion matrix would show which classes are being confused with each other.
